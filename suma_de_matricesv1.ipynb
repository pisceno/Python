```python
import numpy as np

# 1. Definimos la matriz original M (como filas dentro de corchetes)
M = np.array([
    [0, 1, 1, 0, 0],  # Fila A
    [0, 0, 1, 0, 1],  # Fila B
    [0, 0, 0, 1, 0],  # Fila C
    [0, 0, 0, 0, 1],  # Fila D
    [0, 0, 0, 0, 0]   # Fila E
])

# 2. Le pedimos a la computadora que multiplique la matriz por sí misma 3 veces
# En NumPy, la multiplicación de matrices se hace con el símbolo '@'
M3 = M @ M @ M

# 3. Imprimimos el resultado en la terminal
print(M3)
```
